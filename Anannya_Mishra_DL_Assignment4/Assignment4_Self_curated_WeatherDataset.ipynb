Develop an LSTM-based model for time-series forecasting using stock price, weather, or sales datasets.

Dataset used: Self-curated dataset (20 temperature samples)

```python
# Import NumPy, TensorFlow/Keras (LSTM, Dense, Input layers), and MinMaxScaler for preprocessing
import numpy as np
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense
from tensorflow.keras.layers import Input
from sklearn.preprocessing import MinMaxScaler
```

```python
# Sample daily temperature readings (time series)
temps = np.array([25.1, 25.4, 25.8, 26.2, 26.5, 27.0, 27.3, 27.8, 28.2, 28.5,
                  28.9, 29.1, 29.5, 29.8, 30.0, 29.6, 29.2, 28.7, 28.1, 27.5])

# Scale the temperature values to the [0, 1] range for stable LSTM training
scaler = MinMaxScaler(feature_range=(0, 1))
scaled_temps = scaler.fit_transform(temps.reshape(-1, 1))

# Compare original vs scaled values
print("Original Temperatures:")
print(temps)
print("\nScaled Temperatures (MinMaxScaler):")
print(scaled_temps.flatten())
```

```python
# Using 3 past temperature values to predict the next value
dataset = tf.keras.utils.timeseries_dataset_from_array(
    data=scaled_temps[:-1],
    targets=scaled_temps[1:],
    sequence_length=3,
    batch_size=1
)
```

```python
# Define an LSTM-based model: input sequence of 3 timesteps, single feature
model = Sequential([
    Input(shape=(3, 1)),
    LSTM(16),
    Dense(1)
])
# Compile the model with Adam optimizer and mean squared error loss (regression task)
model.compile(optimizer='adam', loss='mse')
```

```python
# Train the LSTM model on the time-series dataset
model.fit(dataset,
          epochs=150,
          verbose=1)

# Took the last 3 known temperature readings to predict the next value
# Scaled the input using the same scaler fitted on the training data
# Predicted the next temperature (in scaled form)
# Converted the prediction back to the original temperature scale

last_three_temps_raw = np.array([28.7, 28.1, 27.5]).reshape(1, 3, 1)
scaled_last_three_temps = scaler.transform(last_three_temps_raw.reshape(-1, 1)).reshape(1, 3, 1)
prediction = model.predict(scaled_last_three_temps,
                           verbose=1)
predicted_original_scale = scaler.inverse_transform(prediction)

print(f"Predicted Next Temperature (scaled): {prediction[0][0]:.4f}")
print(f"Predicted Next Temperature (original scale): {predicted_original_scale[0][0]:.2f}°C")
```
